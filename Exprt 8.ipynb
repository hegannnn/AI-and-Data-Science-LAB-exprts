```python
import seaborn as sn
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.preprocessing import LabelEncoder

iris = sn.load_dataset("iris")
print("\nIris Dataset:\n")
iris
```

```python
print("\nGroup species by size:\n")
iris.groupby('species').size()
```

```python
print("\nGroup species by mean:\n")
iris.groupby('species').mean()
```

```python
print("\nGroup species by sum:\n")
iris.groupby('species').sum()
```

```python
#encode categorical columns
le = LabelEncoder()
iris["species"] = le.fit_transform(iris["species"])

#define target and features
y = iris["petal_width"]
x = iris.drop("petal_width", axis=1)

#split data
xtrain, xtest, ytrain, ytest = train_test_split(x,y,test_size=0.3, random_state=44)

#train model
model = LinearRegression()
p = model.fit(xtrain,ytrain)

#predict and evaluate
xt = p.predict(xtest)
print(f"R2 SCORE(in percentage): {r2_score(ytest, xt)*100:.2f}%")

#predict on new data
new_data = pd.DataFrame({
    "sepal_length" : [5.1],
    "sepal_width" : [3.5],
    "petal_length" : [1.4],
    "species" : [0]
})
xt = p.predict(new_data)

print("\nPREDICTION v/s ACTUAL VALUE: ")
print("Y Prediction: ",xt)
print("Actual Y: ",y[0])
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.preprocessing import LabelEncoder
import pandas as pd
import seaborn as sn

#load and print dataset
data = pd.read_csv(r"C:\Users\hp\Desktop\AIDS\AIDS\Salary Data.csv")
data.dropna(axis=0, inplace=True)
print("SALARY DATASET")
data
```

```python
print("\nGroup job title by size:\n")
data.groupby('Job Title').size()
```

```python
print("\nGroup job title by sum:\n")
data.groupby('Job Title').sum()
```

```python
#encode categorical columns
le = LabelEncoder()
data["Job Title"] = le.fit_transform(data["Job Title"])
data["Education Level"] = le.fit_transform(data["Education Level"])
data["Gender"] = le.fit_transform(data["Gender"])  # Fix for the error

#define target and features
y = data["Salary"]
x = data.drop("Salary", axis=1)

#split data
xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size=0.4, random_state=100)

#train model
model = LinearRegression()
ml = model.fit(xtrain, ytrain)

#predict and evaluate
ypred = ml.predict(xtest)
print(f"R2 SCORE (in percentage): {r2_score(ytest, ypred)*100:.2f}%")

#predict on new data
new_data = pd.DataFrame({
    "Age": [32.0],
    "Gender": [1],  
    "Education Level": [2], 
    "Job Title": [0],        
    "Years of Experience": [5.0]
})
new_pred = ml.predict(new_data)

#print prediction vs actual
print("\nPREDICTION v/s ACTUAL VALUE:")
print("Y Prediction:", new_pred)
print("Actual Y:", y.iloc[0])
```
