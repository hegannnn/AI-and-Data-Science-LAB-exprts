```python
import seaborn as sn
import statistics as st

iris = sn.load_dataset("iris")
iris.dropna(axis=0, inplace=True)

d = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
co = []

print("\nCORRELATION MATRIX\n")
print("\t\tsepal_length\tsepal_width\tpetal_length\tpetal_width")

def mean(l):
    s = 0
    for i in l:
        s += i
    return s / len(l)

def cov(ux, uy, x, y):
    s = 0
    sx2 = 0
    sy2 = 0
    for i in range(len(x)):
        sx = x[i] - ux
        sy = y[i] - uy
        s += sx * sy
        sx2 += sx ** 2
        sy2 += sy ** 2
    deno = (sx2 * sy2) ** 0.5
    return s, deno

for i in range(len(d)):
    x = iris[d[i]]
    r = []
    for j in range(len(d)):
        y = iris[d[j]]
        ux = mean(x)
        uy = mean(y)
        num, deno = cov(ux, uy, x, y)
        c = num / deno
        r.append(round(c, 6))
    co.append(r)

for i in range(len(co)):
    print(f"{d[i]}\t", end="")
    for j in range(len(co)):
        print(f"{co[i][j]:<10}", end="\t")
    print()
```
