```python
import numpy as np
import statistics as st
from scipy.stats import binom, poisson

data = list(map(float, input("Enter data: ").split()))

mean = np.mean(data)
median = np.median(data)
mode = st.multimode(data)
sd = np.std(data, ddof=0)

n = int(input("Enter no. for finding percentile for: "))
percentile = np.percentile(data, n)

print("\nData: ",data)
print("Mean: ",mean)
print("Median: ",median)
print("Mode: ",mode)
print("Standard Deviation: ",sd)
print("Percentile: ",percentile)
```

```python
print("Binomial Distribution:\n")

n = int(input("Enter no. of trials: "))
x = int(input("Enter no. of successes: "))
p = float(input("Enter probability of success: "))
bd = binom.pmf(x,n,p)
print(f"P(X={x}) = {bd:.4f}")
```

```python
print("Poisson Distribution:\n")

x = int(input("Enter x:"))
l = int(input("Enter lambda:"))
pd = poisson.pmf(x,l)
print(f"P(X={x}) = {pd:.4f}")
```

```python
print("Uniform Distribution:\n")

a = int(input("Enter a:"))
b = int(input("Enter b: "))
ud = 1/(b-a)
print(f'f(x) = {ud:.4f}')
```

```python
print("Normal Distribution:\n")

n = int(input("Enter value: "))
m = int(input("Enter mean: "))
sd = int(input("Enter standard deviation: "))

nd = (n-m)/sd
print(f"f(x) = {nd:.4f}")
```
