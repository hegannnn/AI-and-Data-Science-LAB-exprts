```python
# Matrix Addition

#Matrix A
m = int(input("Enter no. of rows for matrix A:"))
n = int(input("Enter no. of columns for matrix B"))
matA = []

print("Enter elements for matrix A:")
for i in range(m):
    row = []
    for j in range(n):
        e = int(input())
        row.append(e)
    matA.append(row)

print("Matrix A: \n",matA)

# Marix B
p = int(input("Enter no. of rows for matrix B:"))
q = int(input("Enter no. of columns for matrix B:"))
matB = []

print("Enter elements for matrix B:")
for i in range(p):
    row = []
    for j in range(q):
        e = int(input())
        row.append(e)
    matB.append(row)

print("Matrix B: \n",matB)

# Sum Matrix
matC = []

if (m!=p or n!=q):
    print("Addition is not possible")
else:
    for i in range(m):
        row = []
        for j in range(n):
            r = matA[i][j] + matB[i][j]
            row.append(r)
        matC.append(row)

    print("Sum matrix: \n",matC)
```

```python
# Matrix Subtraction

#Matrix A
m = int(input("Enter no. of rows for matrix A:"))
n = int(input("Enter no. of columns for matrix B"))
matA = []

print("Enter elements for matrix A:")
for i in range(m):
    row = []
    for j in range(n):
        e = int(input())
        row.append(e)
    matA.append(row)

print("Matrix A: \n",matA)

# Matrix B
p = int(input("Enter no. of rows for matrix B:"))
q = int(input("Enter no. of columns for matrix B:"))
matB = []

print("Enter elements for matrix B:")
for i in range(p):
    row = []
    for j in range(q):
        e = int(input())
        row.append(e)
    matB.append(row)

print("Matrix B: \n",matB)

#Difference matrix
matC = []
if (m!=p or n!=q):
    print("Subtraction not possible")
else:
    for i in range(m):
        row = []
        for j in range(n):
            r = matA[i][j] - matB[i][j]
            row.append(r)
        matC.append(row)

    print("Difference matrix:\n",matC)
```
